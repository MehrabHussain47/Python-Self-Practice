```python
import pandas as pd
import numpy as np
```

```python
data = pd.DataFrame(
    {
        'Name': ['Alice', 'Bob', np.nan, 'Daisy', 'Ethan'],
        'Age' : [25, np.nan, 29, 22, np.nan], # np.nan is for null value
        'salary':[50000, 60000, np.nan, 50000, 62000]
    }
)
```

```python
data.head()
```

```python
data_dropna = data.dropna()
```

```python
data_dropna.head()
```

```python
data_filled = data.fillna(
    {
        'Name': 'Unknown',
        'Age': data['Age'].mean(),
        'salary': 0
    }
)
```

```python
data_filled.head()
```

```python
data_filled['Age'] = data_filled['Age'].astype(int)
```

```python
data_filled.head()
```

```python
data_filled['salary'] = data_filled['salary'].astype(int)
```

```python
data_filled.head()
```

```python
# Join Array
# We can see the vertical join
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6]])
```

```python
a
```

```python
b
```

```python
# Vertical Stacked Join => Vertically Join
v_stacked = np.vstack([a, b])
```

```python
# Dimenstion same hote hobe => column number same hote hobe
v_stacked # a array er row borabor b array join hobe
```

```python
# We can see the horizontal join
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])
```

```python
# Horizontal Join => column borabor join hobe
# Horizontal Join e row number same hote hobe
h_stacked = np.hstack([a, b])
```

```python
h_stacked
```

```python
# We can see the depth wise join
a
```

```python
b
```

```python
# Depth Wise Join => height wise join hishebe kaj kore
# Ekhetre a and b er shape same hote hobe
d_stacked = np.dstack([a, b]) # 2D array to 3D array create hobe
```

```python
# we can see that 1st row of a and 1st row of b mile ekta 2D matrix
# 2nd row of a and 2nd row of b mile 2D matrix.
d_stacked
```
