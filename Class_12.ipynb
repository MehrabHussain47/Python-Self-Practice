# EDA(Exploratory Data Analysis) Visualization

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

np.random.seed(42)  # Data elomelo kore felbe jate similar data porpor nah ashe

districts = ['Downtown', 'East Side', 'West End', 'Northville', 'Southtown']
complaint_types = ['Power Cut', 'Trafic', 'Traffick', 'Noise', 'Garbage', 'Water Leak', 'Power Cut', 'Ice Cream Cart']
ice_cream_flavors = ['Vanilla', 'Chocolate', 'Strawberry', 'Mint', 'Mango', 'Lemon', 'Pistachio', 'Cookie Dough', 'Coffee', 'Blueberry']

n_rows = 500

data = {
    'district': np.random.choice(districts, size=n_rows),
    'complaint_type': np.random.choice(complaint_types, size=n_rows),
    'temperature': np.random.normal(loc=32, scale=5, size=n_rows),
    'electricity_usage': np.random.normal(loc=250, scale=50, size=n_rows),
    'num_ice_cream_carts': np.random.choice([0, 1, 2, 5, 10, 10, 10], size=n_rows),
    'survey_rating': np.random.choice([1, 2, 3, 4, 5, np.nan], size=n_rows),
}

df = pd.DataFrame(data)
```

```python
df.head()
```

```python
df.loc[5:10, 'complaint_type'] = np.nan # 5-10 row te null value dekhabe
df = pd.concat([df, df.iloc[0:5]]) # 0-5 row te repeat kore concat korbo or duplicate value dibo
```

```python
df.head()
```

```python
df.shape  # 5 ta duplicate value concat koray 500 ta row theke 505 ta hoise
```

```python
print(df.isnull().sum()) # complaint_type amra 6ta generate korsi nan value diye
# survey_rating auto generate hoyeche nan value diye randomly
```

```python
print(df.duplicated().sum())  # Ektu age 5 ta duplicate value add korlam
```

```python
print(df.dtypes)  # prottekta column er data type
```

```python
print(df['district'].value_counts())  # ekekta district koybar kore ache
```

```python
complaints_per_district = df.groupby('district')['complaint_type'].count()  # Prottek ta district e koyta complaint ase
print(complaints_per_district)
```

# Histogram

```python
# Town ta te temperature distribution kemon eita dekhbo
sns.set(style="whitegrid")  # Styling korlam darkgrid o dite partam ete grid darker hobe

plt.figure(figsize=(8,4)) # Plot korbo
sns.histplot(df['temperature'], kde=True, bins=10)  # color skyblue deya jabe ar bins 10 bolte 10 vage vag korlam
plt.title('Temperature Distribution in SunnyVille')
plt.xlabel('Temperature Degree')
plt.ylabel('Frequency')
plt.tight_layout()  # axis label, ticks ke overlap hote dey nah
plt.show()
```

# Box Plot

```python
plt.figure(figsize=(10,6))

sns.boxplot(x='district', y='electricity_usage', data=df)
plt.title('Electricity Usage by District')
plt.xlabel('District')
plt.ylabel('Electricity Usage')
plt.tight_layout()
plt.show()  # round shape is outlier
```

# Heatmap

```python
plt.figure(figsize=(8,6))
# number type er correlation hobe ekhane color combination o ache coolwarm bolte blue to red bujay
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='plasma')
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```

# Bar Chart

```python
# Kun dhoroner complaint type e error ache
plt.figure(figsize=(10, 4))
df['complaint_type_clean'].value_counts().plot(kind='bar', color='skyblue')
plt.title("Frequency of Complaint Types")
plt.xlabel("Complaint Type")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

```python
# Bar Chart: Complaint Types (fixed typos)
df['complaint_type_clean'] = df['complaint_type'].replace({
    'Trafic': 'Traffic', # Vul banangula ke thik korbo --> data clean korlam
    'Traffick': 'Traffic'
})
plt.figure(figsize=(10, 4))
df['complaint_type_clean'].value_counts().plot(kind='bar', color='skyblue')
plt.title("Frequency of Complaint Types")
plt.xlabel("Complaint Type")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```
