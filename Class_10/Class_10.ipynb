```python
# Think: SQL-style joins. Match rows based on common columns.
# merging two json files
left = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Pandy', 'Fuzzy', 'Bamboo']
})

right = pd.DataFrame({
    'ID': [1, 2, 4],
    'Hobby': ['Climbing', 'Sleeping', 'Eating']
})
```

```python
print(left)
```

```python
print(right)
```

```python
# Merge on common column ID
merged = pd.merge(left, right, on='ID', how='inner')
print(merged)
```

```python
merged = pd.merge(left, right, on='ID', how='left')
print(merged)
```

```python
merged = pd.merge(left, right, on='ID', how='right')
print(merged)
```

```python
# using join operation
left = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Pandy', 'Fuzzy', 'Bamboo']
}).set_index('ID')

right = pd.DataFrame({
    'ID': [1, 2, 4],
    'Hobby': ['Climbing', 'Sleeping', 'Eating']
}).set_index('ID')
join_data = left.join(right, how = 'inner')
print(join_data)
```

```python
# Think: SQL-style joins. Match rows based on common columns.
# concatenate two json files in row wised
left = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Pandy', 'Fuzzy', 'Bamboo']
})

right = pd.DataFrame({
    'ID': [1, 2, 4],
    'Name': ['Climbing', 'Sleeping', 'Eating']
})
# column name duita same thakte hobe
```

```python
print(left)
```

```python
print(right)
```

```python
concat_data = pd.concat([left, right], axis=0)
print(concat_data)
```

```python
# Think: SQL-style joins. Match rows based on common columns.
# concatenate two json files in column wised
left = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Pandy', 'Fuzzy', 'Bamboo']
})

right = pd.DataFrame({
    'ID_2': [1, 2, 4],
    'Hobby': ['Climbing', 'Sleeping', 'Eating']
})
# column name duita different thakte hobe
```

```python
concat_data = pd.concat([left, right], axis=1)
print(concat_data)
```
