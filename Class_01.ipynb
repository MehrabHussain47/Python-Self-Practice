# 2. Bool

<b>Python also evaluates other data types (like numbers, strings, lists, etc.) as True or False in a
boolean context.</b>
<ol>
<li>0, None, “ ” (empty string), [] (empty list), etc. are considered False.</li>
<li>All other values (non-zero numbers, non-empty strings, lists, etc.) are considered True.</li>
</ol>


```python
a = True
print(a)
print(type(a))
```

```python
a = True
b = False
print(type(a)) # Output: <class 'bool'>
print(type(b)) # Output: <class 'bool'>
```

```python
# Numbers
print(bool(0)) # Output: False (0 is considered False)
print(bool(1)) # Output: True (Non-zero number is considered True)
# Strings
print(bool("")) # Output: False (Empty string is False)
print(bool("Hello")) # Output: True (Non-empty string is True)
# Lists
print(bool([])) # Output: False (Empty list is False)
print(bool([1, 2])) # Output: True (Non-empty list is True)
```

# 5. Tuple
<ol><li>A tuple is a collection of ordered, immutable (unchangeable), and heterogeneous (different data
types) elements.</li>
<li>Tuples are faster than lists because they are immutable.</li>
<li>Tuples are defined using parentheses ().</li></ol>

<p>Use a Tuple (tuple) When:<ol><li>Immutability is needed (data should not change).</li>
<li>Performance
is important (tuples are faster than lists).</li>
<li>Memory optimization is required (tuples use less
memory).</li>
<li>Tuples represent fixed structures like coordinates, database records, or settings.</li>
<li>Tuples can be used as dictionary keys, unlike lists.</li></p>

```python
a = (1, 2, 3)
type(a)
```

```python
a[0:2]
```

```python
# Empty tuple
empty_tuple = ()
# Tuple with elements
numbers = (1, 2, 3, 4)
# Tuple with different data types
mixed_tuple = (1, "hello", 3.14, True)
# Tuple with one element (comma is required!)
single_element_tuple = (5,)
```

# 7. Set
<p>A set is an unordered collection of unique elements.
<ol>
<li>Sets are mutable (can be modified), but they do not allow duplicate values.</li>
<li>Sets are defined using curly braces {} or the set() constructor.</li></ol></p>

```python
a = {'a', 'b', 'c', 'a', 1}
```

```python
type(a)
```

```python
a
```

```python
# Basic Set
fruits = {"apple", "banana", "cherry"}
print(fruits) # Output: {'apple', 'banana', 'cherry'}
```

```python
# Set with Duplicates (Duplicates will be removed)
fruits = {"apple", "banana", "cherry", "apple", "banana"}
print(fruits) # Output: {'apple', 'banana', 'cherry'}
```

```python
# Creating a Set from a List
numbers = set([1, 2, 3, 4, 5])
print(numbers) # Output: {1, 2, 3, 4, 5}
```

<p><b>Set Operations:</b>
<ol><li>add() Adds an elements to the set.</li>
<li>clear() Removes all elements from the set.</li>
<li>copy() Returns a shallow copy of the set.</li>
<li>discard() Removes an element if it exists(does not raise an error).</li>
<li>remove() Removes an element (raises an error if element is not found).</li>
<li>union() Returns a new set with all elements from both sets.</li>
<li>intersection() Returns a new set with elements common to both sets</li>
<li>difference() Returns a new set with elements in the first set but not in the second.</li>
<li>issubset() Checks if the set is a subset of another set.</li>
<li>issuperset() Checks if the set is a superset of another set.</li></ol></p>

```python
# Set Union (|)
set1 = {1, 2, 3}
set2 = {3, 4, 5}
union_set = set1 | set2
print(union_set) # Output: {1, 2, 3, 4, 5}
```

```python
# Set Intersection (&)
set1 = {1, 2, 3}
set2 = {3, 4, 5}
intersection_set = set1 & set2
print(intersection_set) # Output: {3}
```

```python
# Set Difference (-)
set1 = {1, 2, 3}
set2 = {3, 4, 5}
difference_set = set1 - set2
print(difference_set) # Output: {1, 2}
```

# If Statements
<p><b>An if-else statement is used to make decisions in Python. It checks a condition:</b>
<ul><li>If the condition is True, a block of code runs.</li>
<li>If the condition is False, another block of code runs (if else is used).</li></ul></p>


```python
a = 17

if a >= 18:
  print("You are adult!")
else:
  print("You are child!")
```

```python
age = 18
if age >= 18:
  print("You are eligible to vote!")
else:
  print("You are not eligible to vote.")
```

```python
### if-elif-else (Multiple Conditions)
marks = 85
if marks >= 90:
  print("Grade: A+")
elif marks >= 80:
  print("Grade: A")
elif marks >= 70:
  print("Grade: B")
else:
  print("Grade: C")
```

```python
## Nested if-else (Conditions inside Conditions)
num = 10
if num > 0:
  print("Positive number")
  if num % 2 == 0:
    print("Even number")
  else:
    print("Odd number")
else:
  print("Negative number or zero")
```

# Logical Operators
<p><ol><li><b>and</b> Returns True if both conditions are True</li>
<li><b>or</b> Returns True if at least one condition is True</li>
<li><b>not</b> Reverses the result: True → False, False → True</li></ol></p>

```python
x = 7
print(x>5 and x<10)
```

```python
x = 7
print(x>5 and x>10)
```

```python
x = 7
print(x>5 or x>10)
```

```python
x = 7

print(x > 5 and x < 10) # True (because 7 is greater than 5 AND less than 10)
print(x > 5 and x > 10) # False (because second condition is False)
```

```python
x = 3
print(x > 5 or x < 2) # False OR False → False
print(x > 5 or x == 3) # False OR True → True
```

```python
x = 10
print(not(x > 5)) # not(True) → False
print(not(x < 5)) # not(False) → True
```
