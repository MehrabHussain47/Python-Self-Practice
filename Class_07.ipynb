```python
# Set a random seed.
np.random.seed(0)

x1 = np.random.randint(10, size=4)  # one dimension
x2 = np.random.randint(10, size=(3,5))  # two dimension
x3 = np.random.randint(10, size=(3,4,5))  # three dimenstion

print("x1 ndim:", x1.ndim)
print("x1 shape:", x1.shape)
print("x1 size:", x1.size)
```

```python
print("x2 ndim:", x2.ndim)
print("x2 shape:", x2.shape)
print("x2 size:", x2.size)
```

```python
print("x3 ndim:", x3.ndim)
print("x3 shape:", x3.shape)
print("x3 size:", x3.size)
```

# Array Indexing

The important thing to remember is that indexing in python starts at zero.

```python
x1 = np.array([4,3,4,4,8,4])
x1
```

```python
# access value to index zero.
x1[0]
```

```python
# access fifth value
x1[4]
```

```python
# get the last value
x1[-1]
```

```python
# get the second last value
x1[-2]
```

```python
# in a multidimensional array, we need to specify row and column index
x2 = np.array([[3,7,5,5],[0,1,5,9],[3,0,5,0]])
x2
```

```python
# 3rd row and the last column value (4th column) from the 3rd row
x2[2,-1]
```

```python
# replace value at 0,0 index
x2[0,0] = 12
x2
```

# Array Slicing

Now, we will learn to access multiple or a range of elements from an array.

```python
x = np.arange(20)
x
```

```python
# from start to 4th position
x[:5]
```

```python
# from 4th position to end
x[4:]
```

```python
# from 4th to 6th position
x[4:7]
```
